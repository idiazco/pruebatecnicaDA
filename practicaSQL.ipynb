```python
'ejercicio 1'
'listado de clientes de más 110 € en total'
SELECT 
    first_name AS nombre,
    last_name AS apellido,
    SUM(amount) AS total_gastado
'juntando la tabla CUSTOMER a PAYMENT usando la KEY "customer_id'
FROM 
    customer
JOIN
    payment ON customer_id = customer_id
'agrupamos lo gastado por cada cliente'
GROUP BY
    customer_id, first_name, last_name
'y ponemos la condición de que tiene que ser mayor de 110 €'
HAVING
    SUM(amount) > 110
ORDER BY
    total_gastado DESC;
```

```python
'ejercicio 2'
'clientes que nunca hicieron ninguna compra'
SELECT
    first_name AS nombre,
    last_name AS apellido,
    email 'por si los queremos contactar'
FROM 
    customer
JOIN
    payment ON customer_id = customer_id
WHERE
    payment_id IS NULL
```
